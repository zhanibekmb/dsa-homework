````python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
````

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База
# 1.
def to_num(s):
  t = "".join(str(s).split()).replace(",", ".")
  return float(t)

for r in rows:
  r["amount"] = to_num(r["amount"])
  print(r["amount"], type(r["amount"]))
````

````python
CITIES = {"алматы" : "Алматы", "almaty": "Алматы",
          "астана": "Астана", "astana": "Астана",
          "шымкент": "Шымкент", "shymkent": "Шымкент",
          "караганда": "Караганда", "karaganda": "Караганда",
          "актобе": "Актобе", "aktobe": "Актобе",
          "атырау": "Атырау", "atyrau": "Атырау",}

def clean_city(s):
  return CITIES.get(s.lower().strip(), None)

for r in rows:
  r["city"] = clean_city(r["city"])
  print(r["city"])
````

````python
# 2.
categories = {}
for r in rows:
  if r["category"] not in categories:
    categories[r['category']] = 1
  else:
    categories[r['category']] += 1
print(categories)

sorted_categories = dict(sorted(categories.items(), key=lambda kv: kv[1], reverse=True))
print(sorted_categories)
````

````python
# 3.
revenue = {}
count = {}
for r in rows:
  if r["status"] != "оплачен":
    continue
  if r["category"] not in revenue:
    revenue[r["category"]] = r["amount"]
    count[r["category"]] = 1
  else:
    revenue[r["category"]] += r["amount"]
    count[r["category"]] += 1
print(revenue)
print(count)

print(f"{'Категория':<14}{'Заказов':>8}{'Выручка':>14}{'Средний чек':>14}")
for i in sorted(revenue, key=revenue.get, reverse=True):
  print(f"{i:<14}{count[i]:>8}{revenue[i]:>14}{revenue[i] / count[i]:>14.2f}")
````

````python
# 4.
orders_by_month = {}
for r in rows:
  if r["order_date"][:7] not in orders_by_month:
    orders_by_month[r["order_date"][:7]] = 1
  else:
    orders_by_month[r["order_date"][:7]] += 1
print(f"{'Месяц':<10} {'Заказов':<8}")
for i in orders_by_month:
  print(f"{i:<10} {orders_by_month[i]:<8}")
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт
# 5.
revenue_by_month = {}
for r in rows:
  if r["status"] != "оплачен":
    continue
  month = r["order_date"][:7]
  revenue_by_month[month] = revenue_by_month.get(month, 0) + r["amount"]

print(f"{'Месяц':<10} {'Выручка':>10}")
for month in revenue_by_month:
  print(f"{month:<10} {revenue_by_month[month]:>10.2f}")
````

````python
# 6.
canceled_by_city = {}
total = {}
for r in rows:
  total[r["city"]] = total.get(r["city"], 0) + 1
  if r["status"] == "отменён":
    canceled_by_city[r["city"]] = canceled_by_city.get(r["city"], 0) + 1

for c in sorted(total, key=lambda c: canceled_by_city.get(c, 0) / total[c], reverse=True):
    n = canceled_by_city.get(c, 0)
    print(f"{c:<12}{total[c]:>9}{n:>10}{n / total[c] * 100:>7.1f}%")
````

````python
# 7.
rev = {}
for r in rows:
    if r["status"] == "оплачен":
        rev[r["customer_id"]] = rev.get(r["customer_id"], 0) + r["amount"]

top5 = sorted(rev.items(), key=lambda kv: kv[1], reverse=True)[:5]
for i, (c, v) in enumerate(top5, 1):
    print(f"{i}. {c:<8}{v:>14,.0f}")
````

````python
# 8.
customers_by_city = {}
for r in rows:
  if r["city"] not in customers_by_city:
    customers_by_city[r["city"]] = set()
  customers_by_city[r["city"]].add(r["customer_id"])

print(f"{'Город':<12}{'Клиентов':>10}")
for city in sorted(customers_by_city, key=lambda c: len(customers_by_city[c]), reverse=True):
  print(f"{city:<12}{len(customers_by_city[city]):>10}")
````
