8. Доля заказов выходного дня (`сб`, `вс`) в процентах.

````python
weekend = ['сб', 'вс']
weekend_count = 0
for d in days:
  if d in weekend:
    weekend_count += 1
share = weekend_count / len(days) * 100
print(f"Доля заказов выходного дня {share:.1f}")
````

9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
def save(savings):
  months = 0
  while savings * months < 400000:
    months += 1
  print(f"Студент накопит через {months} месяцев")

save(15000)
save(16000)
````

10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.

````python
# Челлендж
unique_cities = []
revenue_by_city = []
for i in range(len(cities_clean)):
  city = cities_clean[i]
  if city not in unique_cities:
    unique_cities.append(city)
    revenue_by_city.append(amounts_int[i])
  if paid[i]:
    idx = unique_cities.index(city)
    revenue_by_city[idx] += amounts_int[i]
unique_cities, revenue_by_city
````

````python
n = len(unique_cities)
for i in range(n-1):
  for j in range(n-1-i):
    if revenue_by_city[j] < revenue_by_city[j+1]:
      revenue_by_city[j], revenue_by_city[j+1] = revenue_by_city[j+1], revenue_by_city[j]
      unique_cities[j], unique_cities[j + 1] = unique_cities[j + 1], unique_cities[j]
print(unique_cities)
print(revenue_by_city)
````

11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.

````python
m = len(amounts_int)
amounts_int_sorted = sorted(amounts_int)
if m % 2 == 0:
  mid1 = amounts_int_sorted[m // 2 - 1]
  mid2 = amounts_int_sorted[m // 2]
  median = (mid1 + mid2) / 2
  print(median)
else:
  median = amounts_int_sorted[m // 2]
  print(median)
````

12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.

````python
week = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]
for day in week:
    total = 0
    for i in range(len(days)):
        if days[i] == day:
            total += amounts_int[i]
    bar = "█" * (total // 5000)
    print(f"{day} | {bar} {total}")
````
